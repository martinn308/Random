# Heat stress checker: converts Celsius to Fahrenheit and compares with the heat stress threshold
# General stress: extended exposure to temperatures above 25 °C (77 °F) with high humidity causes heat stress in humans, crops, and livestock.

```python

THRESHOLD_C = 25  # heat stress threshold in degrees Celsius


def celsius_to_fahrenheit(celsius):
    """Convert a temperature from Celsius to Fahrenheit."""
    return celsius * 9 / 5 + 32


def main():
    # Works out the threshold in Fahrenheit using the same function
    threshold_f = celsius_to_fahrenheit(THRESHOLD_C)
    print(f"Heat stress threshold: {THRESHOLD_C} °C ({threshold_f:.0f} °F)\n")

    # Asks how many temperatures the user wants to check
    count = int(input("How many temperatures do you want to check? "))

    # Loop once for every temperature requested.
    for i in range(1, count + 1):
        celsius = float(input(f"Enter temperature {i} in °C: "))

        # Convert using the function
        fahrenheit = celsius_to_fahrenheit(celsius)
        print(f"{celsius:.1f} °C = {fahrenheit:.1f} °F")

        # Compares with the set threshold and reports the findings
        if celsius > THRESHOLD_C:
            print("ABOVE the threshold: risk of heat stress for humans, crops and animals.\n")
        elif celsius < THRESHOLD_C:
            print("BELOW the threshold: heat stress is unlikely.\n")
        else:
            print("EXACTLY at the threshold: watch conditions closely.\n")


main()
```
